```python
from dotenv import load_dotenv
load_dotenv("./../.env")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = "gemini-3.1-flash-lite-preview"
llm = ChatGoogleGenerativeAI(model = model)
llm.invoke("Hello, introduce me abour yourself")
```

```python
from utils import base_tools
base_tools.web_search.invoke("Why openclaw so hot now?")
```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import TodoListMiddleware
system_prompt = "You are a financial analyst specializing in tech stocks.\nProvide data-driven analysis with clear insights. you have access to web_search tool."
agent_todo = create_agent(
    model=llm,
    tools=[base_tools.web_search],
    system_prompt=system_prompt,
    middleware=[TodoListMiddleware()]
)
```

```python
from langchain_core.messages import HumanMessage
response = agent_todo.invoke({"messages":[HumanMessage("analyze the gold price today and compare it with last year's situation")]})
response
```

```python
for chunk in agent_todo.stream({'messages':["analyze the gold price today and compare it with last year's situation"]}, stream_mode='todos'):
    print(chunk)
    print("------\n")
```
