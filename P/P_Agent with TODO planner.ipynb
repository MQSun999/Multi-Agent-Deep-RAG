```python
def stream_agent_response(agent, query, thread_id="default"):

    config = {"configurable": {"thread_id": thread_id}}

    for chunk in agent.stream(
        {"messages": [HumanMessage(query)]}, stream_mode="messages", config=config
    ):
        # Extract message from chunk
        message = chunk[0] if isinstance(chunk, tuple) else chunk

        # Handle AI messages with tool calls
        if isinstance(message, AIMessage) and message.tool_calls:
            for tool_call in message.tool_calls:
                print(f"\n  Tool Called: {tool_call['name']}")
                print(f"   Args: {tool_call['args']}")
                print()

        # Handle tool responses
        elif isinstance(message, ToolMessage):
            print(f"\n  Tool Result (length: {len(message.text)} chars)")
            print()

        # Handle AI text responses

        elif (
            isinstance(message, AIMessage)
            and message.text
            # 过滤掉tools调用返回的AIMessage
            and chunk[1].get("langgraph_node") != "tools"
        ):
            # Stream the text content
            print(message.text, end="", flush=True)
```

```python
query = "请告诉我工业用火安全的注意事项"
stream_agent_response(agent, query, "session2")
```

```python
from langchain.agents.middleware import SummarizationMiddleware, TodoListMiddleware
```

```python
def get_agent_TODO():
    conn = sqlite3.connect("db/industrial_research_agent_TODO.db", check_same_thread=False)
    checkpointer = SqliteSaver(conn=conn)

    agent = create_agent(
        model=agent_llm,
        checkpointer=checkpointer,
        system_prompt=MULTIMODEL_AGENT_PROMPT,
        tools=[rag_tools.hybrid_search, base_tools.web_search],
        middleware=[
            SummarizationMiddleware(
                model=agent_llm, 
                trigger=[("messages", 25)], 
                keep=("messages", 10)
            ),
            TodoListMiddleware()
        ],
    )

    return agent
```

```python
todo_agent = get_agent_TODO()
todo_agent
```

```python
query = "请告诉我工业用电安全的注意事项，再帮我在网上搜索一下世界各国的常用电压，并总结一下哪个国家的用电比较危险"
stream_agent_response(todo_agent, query, "test2")
```
